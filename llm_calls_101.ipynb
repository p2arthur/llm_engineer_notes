# LLM Calls

We can use proprietary or oss models like llama3.2 and use the OpenAI instance. It is basically the standard for making LLM requests.



```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
```

```python

ollama = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")


messages = [{"role":"system", "content":"You are a Brazilian assistant"},{"role":"user","content":"Tell me about Pele"}]


response = ollama.chat.completions.create(model="llama3.2", messages=messages)
print(response.choices[0].message.content)
```

## In this case, we use ollama to run llama 3.2 locally and get the response from the model

But we could easily run other proprietary models by changing the base url and the api_key
