# 🧠 LLM Calls 101

> **AI Engineer 101 · Day 2** — making our first LLM API calls.

We can use proprietary or open-source models (like **Llama 3.2**) through the `OpenAI` client — it is basically the standard for making LLM requests, and nearly every vendor exposes an OpenAI-compatible endpoint.

**In this notebook**

1. Set up the `OpenAI` client
2. Call a local **Llama 3.2** served by Ollama
3. Swap providers by changing only `base_url` and `api_key`


````python
import os

from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI

load_dotenv()  # loads OPENAI_API_KEY (and friends) from .env, if present
````

## 1 · First call — local Llama 3.2 via Ollama

A chat request is just a list of **messages**, each with a `role` and `content`:

- `system` — sets the assistant's behavior/persona
- `user` — what we're asking

Ollama runs the model locally and exposes an **OpenAI-compatible endpoint** on port `11434`, so the standard `OpenAI` client works out of the box.


````python
ollama = OpenAI(
    base_url="http://localhost:11434/v1",  # Ollama's OpenAI-compatible endpoint
    api_key="ollama",                      # placeholder — Ollama doesn't check it
)

messages = [
    {"role": "system", "content": "You are a Brazilian assistant"},
    {"role": "user", "content": "Tell me about Pelé"},
]

response = ollama.chat.completions.create(
    model="llama3.2",
    messages=messages,
)

display(Markdown(response.choices[0].message.content))
````

Here we used **Ollama** to run Llama 3.2 locally and got the response from the model.

## 2 · Why the `OpenAI` client?

OpenAI pioneered LLM usage through a lightweight endpoint, which made it become the standard:

```python
ollama = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
```

Now all other vendors use the same interface with compatible URLs — so we can easily run other proprietary models by just changing the `base_url` and the `api_key`:

| Provider | `base_url` | Notes |
|---|---|---|
| **Ollama** (local) | `http://localhost:11434/v1` | free, offline, runs on your machine |
| **OpenAI** | _(default)_ | `api_key` read from `OPENAI_API_KEY` |
| **Groq** | `https://api.groq.com/openai/v1` | very fast inference |
| **OpenRouter** | `https://openrouter.ai/api/v1` | one key, many models |
