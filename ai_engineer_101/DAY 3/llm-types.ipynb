# 🧠 LLM Types

> **AI Engineer 101 · Day 3** — the main kinds of LLMs and their most fundamental use cases.

**The four families**

1. **Base models** — raw text completers, the starting point for everything else
2. **Chat models** — interactive assistants, good for content creation
3. **Reasoning models** — chain-of-thought problem solvers
4. **Hybrid models** — chat models with some reasoning, the balanced middle ground


## 1 · Base models

- Trained purely on **next-token prediction** — they don't *answer*, they simply **continue** whatever text you give them.
- Not very powerful on their own, so their main use case is to serve as the **base for training other models** with specific knowledge.

> Ask a base model `What is the capital of France?` and it may well reply `What is the capital of Germany?` — a plausible continuation, not an answer.


## 2 · Chat models

- Base models fine-tuned on conversations + human feedback to **follow instructions**.
- **Good for interactive use cases and content creation** — Q&A, writing, summarizing, translating. This is what we use in almost every LLM call.
- Examples: GPT-4o, Claude, Gemini, `llama3.2` (instruct).


## 3 · Reasoning models

- **Good at chains of thought for problem solving** — math, logic, planning, hard coding.
- Slower and pricier per call, since they "think" at length before answering.
- Examples: OpenAI o-series (o3, o4-mini), DeepSeek R1.

> 🧪 **Funnily enough**, reasoning was discovered when a scientist added a `Wait` to the thought of a chat model — it made the model reconsider and use its previous thoughts as reference, resulting in a much better output.


## 4 · Hybrid models

- A **chat model with some reasoning** built in — a more balanced experience.
- Good at **creating new ideas** (like a chat model) while also **thinking it through** (like a reasoning model).
- Most modern flagships (GPT-5, Claude 4.5) blend both modes.


## Which one to use?

| Type | Follows instructions | Speed / cost | Best for |
|---|---|---|---|
| **Base** | ❌ | fast / cheap | training other models, raw completion |
| **Chat** | ✅ | fast / cheap | interactive use, content creation |
| **Reasoning** | ✅ | slow / pricey | math, logic, problem solving |
| **Hybrid** | ✅ | balanced | a bit of both — ideas + thinking |
