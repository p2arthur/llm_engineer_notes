# Frontier models

Frontier models are the models that are built to push the limits of AI inteligence. Most are proprietary and bring new architectures, new ways of training and are always pushing the boundaries of what is possible with AI

## Frontier models are really good at

Synthesizing information - Answering a question in depth with a structured, well researched answer

Fleshing out a skeleton - Build something more complex from a base. As an example Writing an essay based on a few bullet points 
defined by the operator -> good at creating content

Coding - The ability to design, write and debug code

## They have some limitations too

Very specialized - These frontier models need to be trained on many different kinds of data specific to each field, which makes it very smart on some fields but not as much on others

Lack of recent events - Since these models take a long time and a lot of resources to train, it takes a full cycle in order for engineers to train a model with latest events. Its inteligence is stale until a new model is trained and released

Can confidently make mistakes - These models can say they are very confident about something that is not true
